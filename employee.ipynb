```python
import pandas as pd
import matplotlib.pyplot as plt
df=pd.read_csv('employees.csv')
df
```

```python
plt.figure(figsize=(10,5))
plt.bar(df['Employee_Name'],df['Salary'])
plt.xticks(rotation=45)
plt.xlabel("Employee_names")
plt.ylabel("Salary")
plt.title("Employee_salaries")
plt.tight_layout()
plt.show()
```
