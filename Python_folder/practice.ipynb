```python
import pandas as pd
import numpy as np
df=pd.read_csv("employee_sales.csv")
df.replace("NA",np.nan,inplace=True)
```

```python
df["Age"] = df["Age"].fillna(0)
df.head(10)
```

```python
df["Age"]=df["Age"].replace("Thirty",30)
df["Age"]=pd.to_numeric(df["Age"])
df.head(10)
```

```python
df["City"]=df["City"].fillna(df["City"].mode()[0])
df
```

```python
df["Salary"]=df["Salary"].abs()
df.head(12)
```

```python
df["Joining_Date"] = pd.to_datetime(df["Joining_Date"],format="mixed")
df.head(10)
```

```python
df.head(10)
```

```python
print(df["Monthly_Sales"].isnull().sum())
df
```

```python
df["Performance_Rating"]=df["Performance_Rating"].fillna(0).replace("Excellent",4.5)
df
```

```python
df["Joining_Date"].dtype
```

```python
df.to_csv("employee_sales_cleaned.csv", index=False)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
sns.countplot(data=df,x="Department")
plt.title("Employee Distribution By Department")
plt.xticks(rotation=45)
plt.show()
```

```python
#salary by department...!
sns.barplot(data=df,x='Experience',y='Salary')
plt.title("Experience vs salary")
plt.xticks(rotation=45)
plt.show()
```

```python
#Find the maximum and minimum salary from a list of employee salaries.
max_salary=df["Salary"].max()
min_salary=df["Salary"].min()

print("Maximum salary",max_salary)
print("minimum salary",min_salary)
```
