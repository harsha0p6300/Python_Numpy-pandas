```python
import pandas as pd
import numpy as np
df=pd.read_csv("employee_sales.csv")
df.replace("NA",np.nan,inplace=True)
```

```python
df["Age"] = df["Age"].fillna(0)
df.head(10)
```

```python
df["Age"]=df["Age"].replace("Thirty",30)
df["Age"]=pd.to_numeric(df["Age"])
df.head(10)
```

```python
df["City"]=df["City"].fillna(df["City"].mode()[0])
df
```

```python
df["Salary"]=df["Salary"].abs()
df.head(12)
```

```python
df["Joining_Date"] = pd.to_datetime(df["Joining_Date"],format="mixed")
df.head(10)
```

```python
df.head(10)
```

```python
print(df["Monthly_Sales"].isnull().sum())
df
```

```python
df["Performance_Rating"]=df["Performance_Rating"].fillna(0).replace("Excellent",4.5)
df
```

```python
df["Joining_Date"].dtype
```

```python
df.to_csv("employee_sales_cleaned.csv", index=False)
```
