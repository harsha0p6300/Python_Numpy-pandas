```python
import pandas as pd
df=pd.read_csv('messy_retail_orders.csv')
df
```

```python
#findinf the data type of the specific column
print(df["OrderID"].dtype)
```

```python
df.info()
```

```python
#Changing the data of the particular column
df=df.dropna(subset=["OrderID"])
df["OrderID"]=df["OrderID"].astype(int)
df
```

```python
#finding the duplicates
print(df["Customer Name"].duplicated().sum())
```

```python
#Removing the duplicates
df=df.drop_duplicates(subset=["Customer Name"])
df=df.drop_duplicates(subset=["Email"])
```

```python
df["Product Name"] = df["Product Name"].astype("string")
```

```python
#Converting the customer Name column into title to fix the upper and lower issue
df["Customer Name"]=df["Customer Name"].str.title()
df["City"]=df["City"].str.title()
df["Product Category"]=df["Product Category"].str.title()
df
```

```python
#Replacing the Emails with the invalid emails that's ending with @yahoo.com
df["Email"]=df["Email"].str.replace("yahoo.com","gmail.com",regex=False)
df
```

```python
df=df.dropna()
```

```python
df["Product Category"]=df["Product Category"].str.replace("Electronics","Electronic")
df
```
