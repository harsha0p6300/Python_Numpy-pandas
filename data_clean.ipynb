```python
import pandas as pd
df=pd.read_csv('messy_retail_orders.csv')
df
```

```python
#findinf the data type of the specific column
print(df["OrderID"].dtype)
```

```python
df.info()
```

```python
#Changing the data of the particular column
df=df.dropna(subset=["OrderID"])
df["OrderID"]=df["OrderID"].astype(int)
df
```

```python
#finding the duplicates
print(df["Customer Name"].duplicated().sum())
```

```python
#Removing the duplicates
df=df.drop_duplicates(subset=["Customer Name"])
df=df.drop_duplicates(subset=["Email"])
```

```python
df["Product Name"] = df["Product Name"].astype("string")
```

```python
#Converting the customer Name column into title to fix the upper and lower issue
df["Customer Name"]=df["Customer Name"].str.title()
df["City"]=df["City"].str.title()
df["Product Category"]=df["Product Category"].str.title()
df["Order Status"]=df["Order Status"].str.title()
df
```

```python
#Replacing the Emails with the invalid emails that's ending with @yahoo.com
df["Email"]=df["Email"].str.replace("yahoo.com","gmail.com",regex=False)
df
```

```python
df=df.dropna()
```

```python
df["Product Category"]=df["Product Category"].str.replace("Electronics","Electronic")
df
```

```python
print(df["Product Name"].dtype)
df
```

```python
#Removing the error in the unit price column
df["Unit Price"] = df["Unit Price"].str.replace("â‚¹", "", regex=False)
df.loc[26]
```

```python
df["Quantity"].dtype
```

```python
df["Quantity"]=df["Quantity"].astype(int)
df
```

```python
print(df["Quantity"].value_counts().sort_index())
```

```python
df[df["Quantity"] > 10]
```

```python
df = df[~df["Quantity"].isin([50, 99, 120])]
df
```

```python
df["Discount"] = df["Discount"].str.replace("%", "", regex=False)
df
```

```python
#Converting to the numeric
df["Discount"] = pd.to_numeric(df["Discount"])
df
```

```python
print(df["Discount"].dtype)
```

```python
#adding a new column
df["Discount_Display"] = df["Discount"].astype(str) + "%"
df
```

```python
#Droping the unwanted columns
df = df.drop(columns=["Discount"])
df
```

```python
df["Payment Method"] = df["Payment Method"].str.replace("COD", "Cash on Delivery", regex=False)
df["Payment Method"] = df["Payment Method"].str.replace("cod", "Cash on Delivery", regex=False)
df["Payment Method"] = df["Payment Method"].str.replace("DC", "Debit Card", regex=False)
df["Payment Method"] = df["Payment Method"].str.replace("debit card", "Debit Card", regex=False)
df["Payment Method"] = df["Payment Method"].str.replace("credit card", "CREDIT CARD", regex=False)
df["Payment Method"] = df["Payment Method"].str.replace("NB", "Net Banking", regex=False)
df["Payment Method"] = df["Payment Method"].str.replace("CC", "CREDIT CARD", regex=False)
df
```

```python
#changing the date format
df["Order Date"] = pd.to_datetime(
    df["Order Date"],
    errors="coerce"
)
```

```python
print(df["Order Date"].dtype)
```

```python
df["Order Date"] = pd.to_datetime(
    df["Order Date"],
    format="%d-%m-%Y",
    errors="coerce"
)
df
```
